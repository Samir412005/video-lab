# Chatterbox — Arabic TTS test

Tests **Chatterbox Multilingual V3** (`resemble-ai/chatterbox`, MIT) on one Arabic line, as a
possible upgrade over Piper kareem: more natural prosody + zero-shot voice cloning.
Runs on a free Colab **T4 GPU** (or CPU, ~3x real-time). No login, no payment.

Expected output: `chatterbox_ar.wav` (Arabic voiceover) + its duration.

```python
import torch
import torchaudio
from chatterbox.mtl_tts import ChatterboxMultilingualTTS

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('device:', device)
model = ChatterboxMultilingualTTS.from_pretrained(device=device)
print('sample rate:', model.sr)

text = 'الجزائر، أرض التاريخ والحضارة، من البحر إلى الصحراء.'
wav = model.generate(text, language_id='ar')
torchaudio.save('chatterbox_ar.wav', wav, model.sr)
print('saved chatterbox_ar.wav, duration:', round(wav.shape[-1] / model.sr, 2), 's')
print('(Piper kareem version of the same line: 6.13 s)')
```

## Troubleshooting

- **Install fails on the git pin:** the PyPI fallback in the install cell covers it.
- **Arabic sounds off:** try `language_id='ar'` with a shorter sentence, or pass a short
  reference voice clip via `audio_prompt_path=` for zero-shot cloning.
- **Out of memory on CPU:** the model is 500M params — it fits, but generation is slower;
  keep the text to one or two sentences.
- **Compare with Piper:** the same line with Piper kareem is 6.13 s; judge naturalness
  by listening, not by duration.
